```python
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
print("Total sales by region:\n", region_sales)
top_region = region_sales.idxmax()
print("\nRegion with highest sales:", top_region)

# Step 5: Visualize using a bar chart
plt.figure(figsize=(10,5))
region_sales.plot(kind='bar', color='skyblue')
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
region_sales = df.groupby('City')['Sales'].sum().sort_values(ascending=False)
print("Total sales by region:\n", region_sales)


top_region = region_sales.idxmax()
print("\nRegion with highest sales:", top_region)


plt.figure(figsize=(22,8))
region_sales.plot(kind='bar', color='skyblue')
plt.title('Total Sales by Region')
plt.xlabel('City')
plt.ylabel('Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')

pd.to_datetime(df['Order Date'], format='%d/%m/%2017', errors='coerce')
```

```python
invalid_rows = df[df['Order Date'].isna()]
print(invalid_rows)
```

```python
yearly_sales=df.groupby(df['Order Date'].dt.year)['Sales'].sum()
yearly_sales
```

```python

plt.bar(yearly_sales.index, yearly_sales.values)
plt.xlabel('Year')
plt.ylabel('Total Sales')
plt.title('Year-wise Sales')
plt.show()
```

```python
monthly_sales = (
    df.groupby(pd.Grouper(key='Order Date', freq='M'))['Sales']
      .sum()
      .reset_index()
)
monthly_sales.rename(columns={'Order Date':'Month'}, inplace=True)

print(monthly_sales)
```

```python
plt.figure(figsize=(12,5))
plt.bar(monthly_sales['Month'].astype(str), monthly_sales['Sales'], color='skyblue')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.title('Month-wise Sales')
plt.xticks(rotation=90)
plt.show()
```
